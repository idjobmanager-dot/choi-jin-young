```python
try:
    dan = int(input("출력할 구구단의 단을 입력하세요: "))
    print(f"\n=== {dan}단 ===")
    for i in range(1, 10):
        print(f"{dan} x {i} = {dan * i}")
except ValueError:
    print("올바른 숫자를 입력해주세요.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# 1. 하트 모양 데이터 생성 (매개변수 방정식) - 약간 왼쪽으로 배치
t = np.linspace(0, 2 * np.pi, 1000)
x_heart = 14 * np.sin(t)**3 - 8  # 왼쪽으로 -8 이동
y_heart = 11 * np.cos(t) - 4 * np.cos(2*t) - 1.5 * np.cos(3*t) - 0.8 * np.cos(4*t)

# 그래프 그리기 설정
fig, ax = plt.subplots(figsize=(10, 10))
fig.patch.set_facecolor('#FFF5F5')  # 로맨틱한 베이지 핑크 배경

# 2. 입체 하트 채우기 및 외곽선
ax.plot(x_heart, y_heart, color='#FF2E63', linewidth=3.5, zorder=3)
ax.fill(x_heart, y_heart, color='#FF5E7E', alpha=0.85, zorder=3)  # 메인 핑크
ax.fill(x_heart, y_heart * 0.95, color='#FFA3B1', alpha=0.3, zorder=3)  # 하이라이트

# 3. 큐피트 화살대 (하트를 비스듬히 관통)
x_arrow = np.array([-32, 16])
y_arrow = np.array([-18, 11])
ax.plot(x_arrow, y_arrow, color='#5C3D2E', linewidth=4.5, zorder=2, solid_capstyle='round')

# 입체감 있는 3D 골드 화살촉
head_left = Polygon([[13, 7.5], [19, 13], [15, 9.5]], facecolor='#FFE494', edgecolor='#D4AF37', linewidth=1, zorder=4)
head_right = Polygon([[15, 9.5], [19, 13], [17, 6.8]], facecolor='#D4AF37', edgecolor='#B8860B', linewidth=1, zorder=4)
ax.add_patch(head_left)
ax.add_patch(head_right)

# 레이어드 엔젤 실크 깃털 (화살 뒷부분)
feathers = [
    ([[-29, -16.2], [-32, -14.5], [-31, -16.8]], '#FFFEE0', '#E6C655'),
    ([[-29, -16.2], [-31, -18.5], [-31, -16.8]], '#FFFEE0', '#E6C655'),
    ([[-26.5, -14.8], [-29.5, -13.0], [-28.5, -15.2]], '#FFEAA7', '#D4AF37'),
    ([[-26.5, -14.8], [-28.5, -17.0], [-28.5, -15.2]], '#FFEAA7', '#D4AF37'),
]
for pts, fill_c, edge_c in feathers:
    ax.add_patch(Polygon(pts, facecolor=fill_c, edgecolor=edge_c, linewidth=1, zorder=4))

# 4. 우아하고 예쁜 여성의 옆모습 실루엣 그리기 (오른쪽에 배치)
# 미려한 곡선을 사용하여 이마, 코, 입술, 턱, 목선, 머리카락을 표현
profile_t = np.linspace(-np.pi/2, np.pi, 200)

# 여성 옆모습 라인 정교한 좌표 수치 매핑
profile_x = [
    18, 17.5, 17, 16.8, 16.5, 16.2,  # 이마 상단
    16.1, 16.2, 16.4, 16.5, 16.4,  # 이마에서 콧대로 내려오는 곡선
    15.8, 15.2, 14.7, 14.5,        # 콧대 시작
    14.2, 14.0, 14.1, 14.5, 15.0,  # 코끝과 코 아래 곡선
    15.6, 15.9, 15.5, 15.2, 15.1,  # 윗입술
    15.3, 15.6, 15.4, 15.1, 14.9,  # 아랫입술
    14.9, 15.1, 15.5, 15.9,        # 턱끝
    16.5, 17.2, 18.0, 19.0, 20.2,  # 턱선과 목선
    21.5, 23.0                     # 가슴께로 이어지는 곡선
]

profile_y = [
    15, 13.5, 12, 10.5, 9, 7.5,
    6, 5, 4.2, 3.5, 2.8,
    2.2, 1.8, 1.4, 1.2,
    1.1, 1.4, 1.8, 2.1, 2.3,
    2.4, 2.6, 2.9, 3.1, 3.3,
    3.5, 3.8, 4.1, 4.4, 4.8,
    5.3, 5.8, 6.3, 6.8,
    7.5, 8.3, 9.3, 10.5, 12.0,
    14.0, 16.5
]

# 스케일링 및 위치 미세 조정
px = np.array(profile_x) * 1.2 + 2
py = -np.array(profile_y) * 1.2 + 15

# 우아한 여성 실루엣 드로잉
ax.plot(px, py, color='#4A3742', linewidth=3.5, zorder=5)
# 속을 은은하고 고급스러운 파스텔 바이올렛/로즈 톤으로 채우기
fill_pts = np.column_stack([px, py])
# 오른쪽 캔버스를 닫아 자연스러운 실루엣 채우기 완성
closing_pts = np.array([[33, py[-1]], [33, py[0]]])
full_silhouette = np.vstack([fill_pts, closing_pts])
ax.add_patch(Polygon(full_silhouette, facecolor='#E3D1DB', alpha=0.6, edgecolor=None, zorder=4))

# 우아한 흩날리는 머릿결 장식 라인 추가
hair_x1 = [18*1.2+2, 22*1.2+2, 26*1.2+2, 30*1.2+2]
hair_y1 = [(-15)*1.2+15, (-11)*1.2+15, (-8)*1.2+15, (-7)*1.2+15]
ax.plot(hair_x1, hair_y1, color='#6C5B6B', linestyle='--', alpha=0.7, linewidth=2, zorder=5)

hair_x2 = [17*1.2+2, 21*1.2+2, 25*1.2+2, 28*1.2+2]
hair_y2 = [(-12)*1.2+15, (-8)*1.2+15, (-5)*1.2+15, (-3)*1.2+15]
ax.plot(hair_x2, hair_y2, color='#6C5B6B', alpha=0.5, linewidth=1.5, zorder=5)

# 디스플레이 스타일 정리
ax.set_title("Venus and Cupid's Love", fontsize=22, color='#640D14', fontweight='bold', pad=25)
ax.set_xlim(-33, 33)
ax.set_ylim(-28, 28)
ax.axis('off')

plt.show()
```
